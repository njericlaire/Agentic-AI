### Messages
- Messages are the fundamental unit of context for models in Landchain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with a LLM. Messages are objects that contain:
- **Role** -Identifies the message type (eg system, user)
- **Content** -Represents the actual content of the message (like text,image, audio, documants, etc.)
- **Metadata** -Optional field such as responce information, message IDs and token usage

- LangChain provides a standard message type that works across all models providers, ensuring consistent behaviou regardless of the model being called

```python

import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise ValueError("GROQ_API_KEY is missing. Check your .env file.")

model = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=api_key,
    reasoning_format="hidden",
)

```

```python
model.invoke("Please tell me what is AI")
```

### Message Prompts
- Alternatively, you can pass in a list of messages to the model by providing a list of message objects.
### Message Types
#### System Message
- A System message represents an intial set of instruction that primes the model's behavior. You can use a system message to set the tone, define the model's role and establish guidlines for responses
#### Human Message
- A Human Message represent user input and interactions. They can contain text,images,audio,files and any other amount of multimodal content
#### AI Message
- An AI message represent the output of a modal invocation. They can include multimodal data, tool calls and provider-specific metadata that you can later access
#### Tool Message
- For models that support tool calling. AI messages can contain tool calls. Tool messages are used to pass the result of a single tool execution back to the model.



```python
from langchain.messages import SystemMessage, HumanMessage ,AIMessage

messages = [
    SystemMessage(content="You are a poetry expert."),
    HumanMessage(content="Write a poem about AI")
]

response = model.invoke(messages)
print(response.text)
```

```python
system_message = SystemMessage(content="You are a poetry expert.")
messages = [
    system_message,
    HumanMessage(content="Write a poem about AI")
]
response = model.invoke(messages)
print(response.text)
```

```python
## Detail information to LLM through system message
system_message= SystemMessage("""You are a senior Python developer with extensive experience in writing efficient and maintainable code. Always provide code examples and explain your reasoning. Be concise but thorough in your explanations.""")
messages = [
    system_message,
    HumanMessage(content="Write a Python function to calculate the factorial of a number.")
]
response = model.invoke(messages)
print(response.text)
```

```python
#Message Metadata
human_msg = HumanMessage(content="Hello",
                         name="Alice",
                         id="123")

response = model.invoke([human_msg])
print(response.text)
```

```python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

ai_message = AIMessage(content=[],
                       tool_calls=[
                           {
                               "name": "get_weather",
                               "args": {
                                   "location": "New York"
                               },
                               "id": "123"
                           }
                       ])

weather_result = "Sunny, 72F"
tool_message= ToolMessage(content=weather_result,
                          toot_call_id="123")

message = [HumanMessage("What is the weather in New York ?"),
           ai_message,
           tool_message]
response = model.invoke(message)
print(response.text)
```
