### Streaming.
Most models can stream their output content while it is being generated.By desplaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator thst yields output chunks as they are produced. You can use a loop to process each chunk in real time.


```python
for chunk in model.stream("Why do parrots have colors?"):
    if chunk.content:
        print(chunk.content, end="", flush=True)
```

### Batch
Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can bo done in parrallel.

```python
responses = model.batch(["Why do parrots have colors?", "Why do parrots talk?"])
for response in responses:
    print(response)
```

```python
model.batch(["Why do parrots have colors?", "Why do parrots talk?"],
            config={'max_concurrency': 5}
        )
```
